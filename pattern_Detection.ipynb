```python
import re
```

```python
patterns = [
    r"ignore all previous instructions",
    r"you are now dan",
    r"do anything now"
]
```

```python

def detect_rule_based(prompt, patterns):
    combined_pattern = re.compile(r"\b(" + "|".join(patterns) + r")\b", re.IGNORECASE)


    if combined_pattern.search(prompt):
        print("Potential jailbreak detected!")
```
